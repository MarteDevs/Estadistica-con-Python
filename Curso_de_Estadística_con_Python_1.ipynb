### Paso 1 - Difiniendo el número de clases


---
### 📊 ¿Qué es?

Es una forma de agrupar datos numéricos en **intervalos de igual tamaño** llamados "clases" y contar **cuántos valores caen en cada clase**. Esto facilita visualizar la distribución, detectar patrones y construir gráficos como histogramas.

---

### 🧩 Pasos para construirla

1. **Determinar el número de clases**  
   Se puede usar la regla de Sturges:  
   $$k = 1 + 3.3 \log_{10}(n)$$  
   donde *n* es el número total de datos.

2. **Calcular la amplitud**  
   $$A = \frac{\text{Rango}}{k}$$  
   El rango es:  
   $$\text{Máximo} - \text{Mínimo}$$

3. **Definir los intervalos**  
   Se crean clases como `[10–20), [20–30), ...` con amplitud fija.

4. **Contar la frecuencia**  
   Se registra cuántos datos caen en cada intervalo.

---

### ✅ Ejemplo sencillo

Supón que tienes tiempos de respuesta en segundos:  
`1.2, 0.9, 1.5, 2.1, 1.7, 2.5, 3.0, 2.2, 1.8`

- Min: 0.9  
- Max: 3.0  
- Rango: 2.1  
- Si decides usar 3 clases → amplitud ≈ 0.7  

Entonces los intervalos serían:  
- `[0.9–1.6)`  
- `[1.6–2.3)`  
- `[2.3–3.0)`  

Y luego cuentas cuántos datos hay en cada uno.

---

¿Quieres que hagamos uno con datos reales o simulados? Podemos graficarlo en Python con un histograma bien diseñado.

## <font color=#56B4E9>3.1 Media aritmética</font>
***

Es representada por $\mu$ cuando se refiere a la población y por $\bar{X}$ cuando se refiere a la muestra

# $$\mu = \frac 1n\sum_{i=1}^{n}X_i$$

donde

$n$ = número de observaciones (registros)

$X_i$ = valor de la i-ésima observación (registro)

## - Media (promedio): suma de todos los datos dividida entre la cantidad total.
## Ej.: tiempo promedio de respuesta en una API.

## <font color=#56B4E9>3.2 Mediana</font>
***

Para obtener la mediana de un conjunto de datos, debemos proceder de la siguiente manera:
1. Ordenar el conjunto de datos;
2. Identificar el número de observaciones (registros) del conjunto de datos ($n$);
3. Identicar el elemento del medio:

> Cuando $n$ sea impar, la posición del elemento del medio se obtendrá de la siguiente manera:


# $$Elemento_{Md} = \frac{n+1}{2}$$

> Cuando $n$ sea par, la posición del elemento mediano se obtendrá de la siguiente manera:


# $$Elemento_{Md} = \frac{n}2$$

4. Obtener la mediana:

> Cuando $n$ sea impar:


# $$Md = X_{Elemento_{Md}}$$

> Cuando $n$ sea par:


# $$Md = \frac{X_{Elemento_{Md}} + X_{Elemento_{Md}+1}}2$$
***

## Valor que está justo en el centro cuando los datos están ordenados.
Útil cuando hay valores muy extremos que distorsionan la media.


### Ejemplo 1 - n impar

```python
notas_maria['María'].median()
# esta es la version facil indicando la columna y la mediana
```

***

Las medidas de localización permiten identificar posiciones específicas dentro de una distribución de datos. Son esenciales para entender cómo se distribuyen los valores y realizar comparaciones dentro del conjunto


## <font color=#56B4E9>4.1 Cuartiles, deciles y percentiles</font>
***

Hay una serie de medidas de posición similares en su diseño a la mediana, aunque no son medidas de tendencia central. Como se sabe, la mediana divide la distribución en dos partes iguales en términos del número de elementos en cada parte. Los cuartiles, por otro lado, permiten que la distribución se divida en cuatro partes iguales en términos del número de elementos en cada uno; deciles en diez partes y centiles en cien partes iguales.

📍 1. Cuartiles
Dividen los datos ordenados en cuatro partes iguales.
- Q1 (primer cuartil): el 25% de los datos están por debajo de este valor.
- Q2 (segundo cuartil): es la mediana, representa el 50%.
- Q3 (tercer cuartil): el 75% de los datos están por debajo.
Sirven para detectar concentración de datos y posibles outliers.

🔢 2. Deciles
Dividen los datos en diez grupos iguales.
- D1 a D9 indican el valor debajo del cual se encuentra el 10%, 20%, ... hasta el 90% de los datos.
Son útiles para segmentaciones más finas, como en análisis de ingresos, puntuaciones, etc.

🔟 3. Percentiles
Dividen los datos en 100 partes iguales.
- El percentil 75 indica que el 75% de los valores son menores o iguales a él.
Muy usados en educación (nota de un estudiante respecto al grupo) o salud (percentil de talla o peso infantil).

# <font color=#56B4E9>5 MEDIDAS DE DISPERSIÓN</font>
***

Aunque las medidas de tendencia central y de localización proporcionan un resumen muy importante de los datos, pueden no ser suficientes para caracterizar diferentes conjuntos, especialmente cuando las observaciones de una distribución dada presentan datos muy dispersos.



> Las medidas de dispersión te dicen cuánto se “extienden” los datos alrededor de una medida de localización, como la media. Son clave para entender si tus datos están bien agrupados o si hay mucha variabilidad




## <font color=#56B4E9>5.1 Desviación media absoluta</font>
***
📏 ¿Qué es la Desviación Media Absoluta?
- Es el promedio de las distancias absolutas entre cada dato y la media.
- Te dice, en promedio, cuánto se alejan tus datos del valor medio, sin importar si es por exceso o defecto.

🔍 ¿Por qué es útil?
- Menos sensible a outliers que la varianza.
- Intuitiva: mide distancia promedio en las mismas unidades que la variable.
- Ideal para cuando quieres un resumen robusto pero interpretable.


# $$DM = \frac 1n\sum_{i=1}^{n}|X_i-\bar{X}|$$

## <font color=#56B4E9>5.2 Varianza</font>
***

### Varianza

La varianza se construye a partir de las diferencias entre cada observación y la media de los datos, es decir, la desviación alrededor de la media. Al calcular la varianza, las desviaciones alrededor de la media son elevadas al cuadrado.

### Varianza de la población

# $$\sigma^2 = \frac 1n\sum_{i=1}^{n}(X_i-\mu)^2$$

### Varianza de la muestra

# $$S^2 = \frac 1{n-1}\sum_{i=1}^{n}(X_i-\bar{X})^2$$

```python
notas_maria_dispersion
```

```python
notas_maria_dispersion['Desviacion_cuadrada'] = notas_maria_dispersion['desviacion_absoluta'].pow(2)
```

```python
notas_maria_dispersion
```

```python
notas_maria_dispersion['Desviacion_cuadrada'].sum()/(len(notas_maria_dispersion)-1)
# nuestra varianza
```

```python
# forma rapida
notas_maria_dispersion['María'].var()
# var indica la varianza
```

## <font color=#56B4E9>5.3 Desviación estándar</font>
***

Una de las restricciones de la varianza es el hecho de que proporciona medidas cuadráticas de las unidades originales; la varianza de las medidas de longitud, por ejemplo, está en unidades de área. Por lo tanto, el hecho de que las unidades sean diferentes dificulta la comparación de la dispersión con las variables que la definen. Una forma de eliminar esta dificultad es considerar su raíz cuadrada.

### Desviación estándar de la población

# $$\sigma = \sqrt{\frac 1n\sum_{i=1}^{n}(X_i-\mu)^2} \Longrightarrow \sigma = \sqrt{\sigma^2}$$


### Desviación estándar de la muestra

# $$S = \sqrt{\frac 1{n-1}\sum_{i=1}^{n}(X_i-\bar{X})^2} \Longrightarrow S = \sqrt{S^2}$$

```python
varianza = notas_maria_dispersion['María'].var()
varianza
```
